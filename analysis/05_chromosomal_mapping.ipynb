# Where are these genes?
Is it possible that the signficiant genes are clustered together in the chromosome? It would be cool to find out.

```python
import pandas as pd
import matplotlib.pyplot as plt

'''loading the annotation file'''

annotation_file = pd.read_csv('../data/GSE2034_family_platform_GPL96_annotation.csv', sep=",")
annotation_df = pd.DataFrame(annotation_file)


annotation_df.head(10)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

'''loading the annotation file'''

differential_result = pd.read_csv('../data/differential_expression_results.csv', sep=",")
de_results = pd.DataFrame(differential_result)


de_results.head(10)
```

```python
annotation_subset = annotation_df[
    ["ID", "Gene Symbol", "ENTREZ_GENE_ID"]
].copy()

annotation_subset = annotation_subset.rename(columns={
    "ID": "Probe_ID",
    "Gene Symbol": "Gene_Symbol",
    "ENTREZ_GENE_ID": "Entrez_ID"
})

print("Total probes:", len(annotation_subset))

print("Probes with gene symbols:",
      annotation_subset["Gene_Symbol"].notna().sum())

print("Probes with Entrez IDs:",
      annotation_subset["Entrez_ID"].notna().sum())

print("Probes missing gene symbols:",
      annotation_subset["Gene_Symbol"].isna().sum())

print("Probes missing Entrez IDs:",
      annotation_subset["Entrez_ID"].isna().sum())

multiple_symbols = annotation_subset[
    annotation_subset["Gene_Symbol"].astype(str).str.contains("///", na=False)
]

print("Probes with multiple gene symbols:", len(multiple_symbols))

annotation_subset.head(10)
```

```python
mapped_results = de_results.merge(
    annotation_subset,
    on="Probe_ID",
    how="left"
)

print("Differential expression probes:", len(de_results))
print("After annotation merge:", len(mapped_results))
print("Non-significant probes:", (~de_results["Significant"]).sum())

mapped_results.head()
```

```python
# Check for missing Entrez IDs
print("Missing Entrez IDs:",
      mapped_results["Entrez_ID"].isna().sum())

# Check for probes with multiple Entrez IDs
multiple_entrez = mapped_results[
    mapped_results["Entrez_ID"].astype(str).str.contains("///", na=False)
]

print("Probes with multiple Entrez IDs:", len(multiple_entrez))

# Check number of unique Entrez IDs
print("Unique Entrez IDs:",
      mapped_results["Entrez_ID"].nunique())
```

```python
# Keep probes with a single Entrez ID
unique_entrez = mapped_results[
    mapped_results["Entrez_ID"].notna() &
    ~mapped_results["Entrez_ID"].astype(str).str.contains("///")
].copy()

# Convert Entrez IDs to strings
unique_entrez["Entrez_ID"] = (
    unique_entrez["Entrez_ID"].astype(str).str.strip()
)

print("Probes with unambiguous Entrez IDs:", len(unique_entrez))
print("Unique Entrez IDs:", unique_entrez["Entrez_ID"].nunique())
```

```python
'''Importing and reading Genome GRCh38.p14 data to map locations '''

import pandas as pd


geo_anno = pd.read_csv('../data/genomic.gff', sep="\t", comment="#", header=None)
geo_anno_df = pd.DataFrame(geo_anno)

geo_anno_df.columns = ["seqid", "source", "type", "start", "end", "score", "strand", "phase", "attributes"]

genes_anno = geo_anno_df[geo_anno_df["type"] == "gene"].copy()

genes_anno["Entrez_ID"] = genes_anno["attributes"].str.extract(r"GeneID:(\d+)")
genes_anno["Gene_Symbol"] = genes_anno["attributes"].str.extract(r"gene=([^;]+)")

genes_anno.head(10)
#geo_anno_df.head(50)
```

```python
chromosome_map = {
    f"NC_{i:06d}": str(i)
    for i in range(1, 23)
}

chromosome_map.update({
    "NC_000023": "X",
    "NC_000024": "Y"
})

genes_anno["Chromosome"] = (
    genes_anno["seqid"]
    .str.extract(r"(NC_\d+)", expand=False)
    .map(chromosome_map)
)

genes_anno[["seqid", "Chromosome", "start", "end", "Entrez_ID", "Gene_Symbol"]].head(10)
```

```python
# Keep only the columns needed for mapping
gene_coordinates = genes_anno[["Entrez_ID", "Gene_Symbol", "seqid", "start", "end", "strand"]].copy()

# Remove duplicate Entrez IDs
gene_coordinates = gene_coordinates.drop_duplicates(
    subset="Entrez_ID"
)

print("Unique annotated genes:", gene_coordinates["Entrez_ID"].nunique())
print("Total annotation records:", len(gene_coordinates))
```

```python
mapped_results["Entrez_ID"] = mapped_results["Entrez_ID"].astype(str)
gene_coordinates["Entrez_ID"] = gene_coordinates["Entrez_ID"].astype(str)

chromosome_results = mapped_results.merge(
    gene_coordinates,
    on="Entrez_ID",
    how="left",
    suffixes=("", "_annotation")
)

print("Original probes:", len(mapped_results))
print("After genomic mapping:", len(chromosome_results))
print("Probes with genomic coordinates:", chromosome_results["start"].notna().sum())
print("Probes without genomic coordinates:", chromosome_results["start"].isna().sum())
```

```python
unmapped = chromosome_results[
    chromosome_results["start"].isna()
]

print("Unmapped probes:", len(unmapped))
print("Missing Entrez IDs:", unmapped["Entrez_ID"].isna().sum())
print("Unique unmapped Entrez IDs:", unmapped["Entrez_ID"].nunique())
```

```python
unmapped_ids = set(unmapped["Entrez_ID"].unique())
annotation_ids = set(gene_coordinates["Entrez_ID"].unique())

missing_ids = unmapped_ids - annotation_ids

print("Unique unmapped IDs:", len(unmapped_ids))
print("IDs absent from GFF3:", len(missing_ids))
print("IDs found in GFF3:", len(unmapped_ids & annotation_ids))
```

```python
print("Unique mapped Entrez IDs:",
      chromosome_results["Entrez_ID"].dropna().nunique())

print("Unique mapped gene symbols:",
      chromosome_results["Gene_Symbol_annotation"].dropna().nunique())
```

```python
'''we have duplicate probes matching to the same entrez id. we should confirm that the probes are showing the same expression direction'''

probe_counts = (
    chromosome_results.dropna(subset=["Entrez_ID"])
    .groupby("Entrez_ID")["Probe_ID"]
    .nunique()
)

print("Genes represented by multiple probes:",
      (probe_counts > 1).sum())

print("Maximum probes per gene:", probe_counts.max())

print("Genes represented by one probe:",
      (probe_counts == 1).sum())
```

```python
'''do multiple genes agree?'''
import numpy as np

duplicate_probes = chromosome_results[
    chromosome_results["Entrez_ID"].isin(
        probe_counts[probe_counts > 1].index
    )
].copy()

duplicate_probes["Direction"] = np.where(
    duplicate_probes["Mean_Difference"] > 0,
    "Higher in relapse",
    "Higher in no relapse"
)

direction_agreement = (
    duplicate_probes.groupby("Entrez_ID")["Direction"]
    .nunique()
)

print("Genes with consistent direction:",
      (direction_agreement == 1).sum())

print("Genes with conflicting directions:",
      (direction_agreement > 1).sum())
```

```python
'''statistically significant probes only'''

chromosome_results["Direction"] = np.where(
    chromosome_results["Mean_Difference"] > 0,
    "Higher in relapse",
    "Higher in no relapse"
)

significant_probes = chromosome_results[
    (chromosome_results["Significant"] == True) &
    (chromosome_results["start"].notna())
].copy()

significant_direction_agreement = (
    significant_probes.groupby("Entrez_ID")["Direction"]
    .nunique()
)

print("Genes with significant probes:",
      significant_direction_agreement.size)

print("Consistent significant direction:",
      (significant_direction_agreement == 1).sum())

print("Conflicting significant directions:",
      (significant_direction_agreement > 1).sum())
```

```python
'''Only 1 gene is conflicting in direction within the significant genes, let's inspect'''

conflicting_genes = significant_direction_agreement[
    significant_direction_agreement > 1
]

conflicting_probes = significant_probes[
    significant_probes["Entrez_ID"].isin(
        conflicting_genes.index
    )
]

conflicting_probes[
    ["Probe_ID", "Entrez_ID", "Gene_Symbol",
     "Mean_Difference", "Adjusted_P_Value", "Direction"]]

'''This gene is significant in both relapse and relapse free, we should add in another category for "conflicting direction"'''
```

```python
# Create a chromosome accession-to-number mapping
chromosome_map = {
    f"NC_{i:06d}": str(i)
    for i in range(1, 23)
}
chromosome_map.update({
    "NC_000023": "X",
    "NC_000024": "Y"
})

# Prepare one genomic record per Entrez ID
gene_level_results = gene_coordinates.copy()

gene_level_results["Entrez_ID"] = (
    gene_level_results["Entrez_ID"].astype(str)
)

gene_level_results["Chromosome"] = (
    gene_level_results["seqid"]
    .str.extract(r"(NC_\d+)", expand=False)
    .map(chromosome_map)
)

# Keep genes represented in our differential-expression results
mapped_ids = chromosome_results["Entrez_ID"].dropna().astype(str).unique()

gene_level_results = gene_level_results[
    gene_level_results["Entrez_ID"].isin(mapped_ids) &
    gene_level_results["Chromosome"].notna()
].copy()

# Recalculate direction for significant probes
sig = chromosome_results[
    (chromosome_results["Significant"] == True) &
    (chromosome_results["start"].notna())
].copy()

sig["Direction"] = np.where(
    sig["Mean_Difference"] > 0,
    "Higher in relapse",
    "Higher in no relapse"
)

# Summarize significant probes at gene level
gene_status = (
    sig.groupby("Entrez_ID")
    .agg(
        Significant_Probes=("Probe_ID", "nunique"),
        Direction_Count=("Direction", "nunique"),
        Direction=("Direction", "first")
    )
    .reset_index()
)

gene_status["Entrez_ID"] = gene_status["Entrez_ID"].astype(str)

# Merge expression status with genomic coordinates
gene_level_results = gene_level_results.merge(
    gene_status,
    on="Entrez_ID",
    how="left"
)

# Assign the four visualization categories
gene_level_results["Expression_Category"] = np.select(
    [
        gene_level_results["Direction_Count"] > 1,
        gene_level_results["Direction"] == "Higher in relapse",
        gene_level_results["Direction"] == "Higher in no relapse"
    ],
    [
        "Conflicting directions",
        "Higher in relapse",
        "Higher in no relapse"
    ],
    default="Not significant"
)

print(gene_level_results["Expression_Category"].value_counts())

print("Total rows:", len(gene_level_results))
print("Unique Entrez IDs:", gene_level_results["Entrez_ID"].nunique())
print("Duplicate Entrez IDs:",
      gene_level_results["Entrez_ID"].duplicated().sum())

print("\nGenes per chromosome:")
print(gene_level_results["Chromosome"].value_counts())
```

```python
primary_chromosomes = genes_anno[
    genes_anno["seqid"].str.extract(r"(NC_\d+)", expand=False).isin(chromosome_map)
].copy()

print("Primary chromosome records:", len(primary_chromosomes))
print("Unique Entrez IDs on primary chromosomes:",
      primary_chromosomes["Entrez_ID"].nunique())

print("Entrez IDs with multiple primary chromosome records:",
      primary_chromosomes["Entrez_ID"].duplicated().sum())
```

```python
multiple_locations = (
    primary_chromosomes
    .groupby("Entrez_ID")
    .agg(
        Record_Count=("Entrez_ID", "size"),
        Chromosome_Count=("seqid", "nunique"),
        Start_Count=("start", "nunique"),
        End_Count=("end", "nunique")
    )
    .query("Record_Count > 1")
)

print(multiple_locations)
```

```python
'''how many of the genes are ambiguous locations'''

ambiguous_ids = multiple_locations.index.astype(str)

print(
    "Ambiguous genes in visualization:",
    gene_level_results["Entrez_ID"].isin(ambiguous_ids).sum()
)
```

```python
ambiguous_ids = multiple_locations.index.astype(str)

gene_level_results_clean = gene_level_results[
    ~gene_level_results["Entrez_ID"].isin(ambiguous_ids)
].copy()

print("Genes retained:", len(gene_level_results_clean))
print("Unique Entrez IDs:", gene_level_results_clean["Entrez_ID"].nunique())
```

```python
'''plotting'''

chromosome_lengths = (
    primary_chromosomes
    .assign(
        Chromosome=lambda df: (
            df["seqid"]
            .str.extract(r"(NC_\d+)", expand=False)
            .map(chromosome_map)
        )
    )
    .dropna(subset=["Chromosome"])
    .groupby("Chromosome")["end"]
    .max()
)

print(chromosome_lengths)
```

```python
chromosome_order = [str(i) for i in range(1, 23)] + ["X", "Y"]

gene_level_results_clean["Chromosome"] = pd.Categorical(
    gene_level_results_clean["Chromosome"],
    categories=chromosome_order,
    ordered=True
)

print(gene_level_results_clean["Chromosome"].value_counts().sort_index())

print(gene_level_results_clean["Expression_Category"].value_counts())
```

```python
gene_hover = (
    chromosome_results[
        (chromosome_results["Significant"] == True) &
        (chromosome_results["start"].notna())
    ]
    .groupby("Entrez_ID")
    .agg(
        Mean_Difference=("Mean_Difference", "median"),
        Adjusted_P_Value=("Adjusted_P_Value", "min"),
        Significant_Probe_Count=("Probe_ID", "nunique")
    )
    .reset_index()
)

gene_hover["Entrez_ID"] = gene_hover["Entrez_ID"].astype(str)

gene_level_results_clean = gene_level_results_clean.merge(
    gene_hover,
    on="Entrez_ID",
    how="left"
)

print(gene_level_results_clean[
    ["Gene_Symbol", "Chromosome", "start", "Expression_Category",
     "Mean_Difference", "Adjusted_P_Value"]
].head())
```

```python

import numpy as np
import pandas as pd
import plotly.graph_objects as go

# Prepare genes
plot_data = gene_level_results_clean.sort_values(
    ["Chromosome", "start"]
).reset_index(drop=True)

# Split genes between the two sister chromatids
gene_groups = [
    plot_data.iloc[::2].copy(),
    plot_data.iloc[1::2].copy()
]

# Define the two chromatid paths
t = np.linspace(0, 1, 100)

paths = [
    np.column_stack((-1 + 2*t, 1 - 2*t, 0.08*np.sin(t*np.pi))),
    np.column_stack((-1 + 2*t, -1 + 2*t, 0.08*np.sin(t*np.pi)))
]

# Build a 3D tapered tube around a path
def make_chromatid(path, radius=0.13, sides=32):

    tangent = np.gradient(path, axis=0)
    tangent = tangent / np.linalg.norm(
        tangent, axis=1
    )[:, None]

    reference = np.array([0, 0, 1])

    normal = np.cross(tangent, reference)
    normal = normal / np.linalg.norm(
        normal, axis=1
    )[:, None]

    binormal = np.cross(tangent, normal)

    vertices = []

    for i, point in enumerate(path):

        # Narrow at centromere, rounded toward the ends
        taper = 0.65 + 0.35 * np.sin(np.pi * t[i])
        pinch = 0.52 + 0.48 * min(1, abs(t[i] - 0.5) * 4)
        r = radius * taper * pinch

        for angle in np.linspace(
            0, 2*np.pi, sides, endpoint=False
        ):
            vertex = (
                point
                + r * np.cos(angle) * normal[i]
                + r * np.sin(angle) * binormal[i]
            )
            vertices.append(vertex)

    vertices = np.array(vertices)

    faces_i, faces_j, faces_k = [], [], []
    face_colors = []

    bands = ["#52616B", "#89969D"]

    for i in range(len(path)-1):
        for j in range(sides):

            a = i*sides + j
            b = i*sides + (j+1) % sides
            c = (i+1)*sides + j
            d = (i+1)*sides + (j+1) % sides

            faces_i.extend([a, b])
            faces_j.extend([c, c])
            faces_k.extend([b, d])

            band = bands[(i // 8) % 2]
            face_colors.extend([band, band])

    return vertices, faces_i, faces_j, faces_k, face_colors


fig = go.Figure()

# Draw the two chromatids
for path in paths:

    vertices, i, j, k, colors = make_chromatid(path)

    fig.add_trace(go.Mesh3d(
        x=vertices[:, 0],
        y=vertices[:, 1],
        z=vertices[:, 2],
        i=i,
        j=j,
        k=k,
        facecolor=colors,
        opacity=1,
        flatshading=False,
        lighting=dict(
            ambient=0.45,
            diffuse=0.8,
            specular=0.6,
            roughness=0.45
        ),
        hoverinfo="skip",
        showlegend=False
    ))

# Gene colours
color_map = {
    "Not significant": "#D3D3D3",
    "Higher in relapse": "#E63946",
    "Higher in no relapse": "#2878B5",
    "Conflicting directions": "#8E44AD"
}

# Place gene markers along the chromatids
for arm_number, subset in enumerate(gene_groups):

    subset = subset.reset_index(drop=True)
    positions = np.linspace(0.06, 0.94, len(subset))

    marker_x, marker_y, marker_z = [], [], []

    for position in positions:

        index = int(position * (len(t)-1))
        point = paths[arm_number][index]

        # Put markers slightly outside the chromatid surface
        marker_x.append(point[0] + 0.11)
        marker_y.append(point[1])
        marker_z.append(point[2] + 0.10)

    subset["Plot_X"] = marker_x
    subset["Plot_Y"] = marker_y
    subset["Plot_Z"] = marker_z

    for category, color in color_map.items():

        selected = subset[
            subset["Expression_Category"] == category
        ]

        fig.add_trace(go.Scatter3d(
            x=selected["Plot_X"],
            y=selected["Plot_Y"],
            z=selected["Plot_Z"],
            mode="markers",
            name=category,
            marker=dict(
                size=3.5,
                color=color,
                opacity=0.95
            ),
            customdata=selected[
                [
                    "Gene_Symbol",
                    "Chromosome",
                    "start",
                    "Mean_Difference",
                    "Adjusted_P_Value"
                ]
            ],
            hovertemplate=(
                "<b>%{customdata[0]}</b><br>"
                "Chromosome: %{customdata[1]}<br>"
                "Position: %{customdata[2]:,}<br>"
                "Mean difference: %{customdata[3]:.4f}<br>"
                "Adjusted p-value: %{customdata[4]:.4e}"
                "<extra></extra>"
            )
        ))

# Centromere
fig.add_trace(go.Scatter3d(
    x=[0],
    y=[0],
    z=[0],
    mode="markers",
    marker=dict(
        size=8,
        color="#303840",
        symbol="circle"
    ),
    name="Centromere",
    hoverinfo="skip"
))

fig.update_layout(
    title="3D Chromosome: Relapse-Associated Gene Expression",
    template="plotly_dark",
    width=1000,
    height=900,
    scene=dict(
        xaxis=dict(visible=False),
        yaxis=dict(visible=False),
        zaxis=dict(visible=False),
        aspectmode="data",
        camera=dict(
            eye=dict(x=1.5, y=1.5, z=1.3)
        )
    ),
    legend_title="Expression Category"
)

fig.show()
```
