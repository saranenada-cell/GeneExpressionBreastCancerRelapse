```python
import gzip
import shutil
from pathlib import Path

# Exact source folder
source_dir = Path.cwd() / "cel" / "compressed"

# Destination folder
output_dir = Path.cwd() / "cel" / "extracted"
output_dir.mkdir(parents=True, exist_ok=True)

# Find compressed files
compressed_files = list(source_dir.glob("*.CEL.gz"))

# Extract
for file in compressed_files:
    output_file = output_dir / file.name[:-3]

    with gzip.open(file, "rb") as compressed:
        with open(output_file, "wb") as extracted:
            shutil.copyfileobj(compressed, extracted)

print("Compressed files:", len(compressed_files))
print("Extracted files:", len(list(output_dir.glob("*.CEL"))))
```
